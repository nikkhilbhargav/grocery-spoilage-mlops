# Basic EDA - Grocery Spoilage Dataset

Just checking that the synthetic data makes sense before training models.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/grocery_data.csv')
df.head()
```

```python
df.describe()
```

```python
# how many rows per category
df['category'].value_counts()
```

```python
# spoilage risk % per category - fragile items should have the highest risk
df.groupby('category')['spoilage_risk'].mean()
```

```python
# does higher temperature correlate with more spoilage?
plt.scatter(df['temperature'], df['spoilage_days'], alpha=0.2, s=8)
plt.xlabel('temperature (C)')
plt.ylabel('spoilage_days')
plt.title('Temperature vs Spoilage Days')
plt.show()
```
